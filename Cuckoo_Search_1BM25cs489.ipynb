```python
import numpy as np

def fitness(x):
    return (x[0]-0.01)**2 + (x[1]-64)**2/10000 + (x[2]-3)**2 + (x[3]-128)**2/10000

def cuckoo_search(n=5, Pa=0.25, MaxIter=20):

    nests = np.random.uniform(
        [0.001, 32, 1, 32],
        [0.1, 128, 5, 256], (n, 4)
    )

    scores = np.array([fitness(x) for x in nests])

    best = nests[np.argmin(scores)].copy()

    for t in range(MaxIter):

        i = np.random.randint(n)
        new = nests[i] + 0.01 * np.random.standard_cauchy(4)
        new = np.clip(new, [0.001, 32, 1, 32],
                      [0.1, 128, 5, 256])

        new_score = fitness(new)

        j = np.random.randint(n)

        if new_score < scores[j]:
            nests[j] = new
            scores[j] = new_score

        count = max(1, int(Pa*n))
        worst = np.argsort(scores)[-count:]

        for k in worst:
            nests[k] = np.random.uniform(
                [0.001, 32, 1, 32],
                [0.1, 128, 5, 256]
            )
            scores[k] = fitness(nests[k])

        best = nests[np.argmin(scores)].copy()

        print("Iteration:", t+1, "Best Fitness:", fitness(best))

    return best

best = cuckoo_search()

print("\nOptimized Hyperparameters")
print("Learning Rate:", best[0])
print("Batch Size:", int(best[1]))
print("Number of Layers:", int(best[2]))
print("Hidden Units:", int(best[3]))
```
