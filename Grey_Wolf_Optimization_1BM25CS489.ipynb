```python
import numpy as np

# Objective function
def fitness(x):
    return 1000-(x-35)**2

# Parameters
n_wolves=5
max_iter=50
dim=1
lb=0
ub=90

# Initialize wolf population randomly
wolves=np.random.uniform(lb,ub,(n_wolves,dim))

# Initialize Alpha, Beta and Delta
alpha_pos=np.zeros(dim)
beta_pos=np.zeros(dim)
delta_pos=np.zeros(dim)

alpha_score=-float("inf")
beta_score=-float("inf")
delta_score=-float("inf")

# Main GWO loop
for t in range(max_iter):

    # Calculate fitness
    for i in range(n_wolves):
        score=fitness(wolves[i][0])

        if score>alpha_score:
            delta_score=beta_score
            delta_pos=beta_pos.copy()

            beta_score=alpha_score
            beta_pos=alpha_pos.copy()

            alpha_score=score
            alpha_pos=wolves[i].copy()

        elif score>beta_score:
            delta_score=beta_score
            delta_pos=beta_pos.copy()

            beta_score=score
            beta_pos=wolves[i].copy()

        elif score>delta_score:
            delta_score=score
            delta_pos=wolves[i].copy()

    # Decrease parameter a from 2 to 0
    a=2-2*t/max_iter

    # Update wolf positions
    for i in range(n_wolves):
        for j in range(dim):

            r1,r2=np.random.rand(),np.random.rand()
            A1=2*a*r1-a
            C1=2*r2
            D_alpha=abs(C1*alpha_pos[j]-wolves[i][j])
            X1=alpha_pos[j]-A1*D_alpha

            r1,r2=np.random.rand(),np.random.rand()
            A2=2*a*r1-a
            C2=2*r2
            D_beta=abs(C2*beta_pos[j]-wolves[i][j])
            X2=beta_pos[j]-A2*D_beta

            r1,r2=np.random.rand(),np.random.rand()
            A3=2*a*r1-a
            C3=2*r2
            D_delta=abs(C3*delta_pos[j]-wolves[i][j])
            X3=delta_pos[j]-A3*D_delta

            wolves[i][j]=(X1+X2+X3)/3

            # Keep position within limits
            wolves[i][j]=np.clip(wolves[i][j],lb,ub)

# Display result
print("Grey Wolf Optimization Result")
print("Optimal Tilt Angle:",round(alpha_pos[0],4),"degrees")
print("Maximum Fitness:",round(alpha_score,4))
```
