```python
import random
import math

locations={
    "Depot":(0,0),
    "A":(2,3),
    "B":(5,4),
    "C":(6,1),
    "D":(3,6),
    "E":(8,3)
}

num_ants=10
iterations=50
alpha=1
beta=2
rho=0.5
Q=100

nodes=list(locations.keys())
pheromone={}
distance={}

for i in nodes:
    for j in nodes:
        if i!=j:
            d=math.dist(locations[i],locations[j])
            distance[i,j]=d
            pheromone[i,j]=1.0

best_route=None
best_distance=float('inf')

for iteration in range(iterations):
    all_routes=[]

    for ant in range(num_ants):
        unvisited=nodes.copy()
        unvisited.remove("Depot")

        route=["Depot"]
        current="Depot"

        while unvisited:
            probabilities=[]
            total=0

            for next_node in unvisited:
                p=(pheromone[current,next_node]**alpha)*((1/distance[current,next_node])**beta)
                probabilities.append(p)
                total+=p

            probabilities=[p/total for p in probabilities]

            next_node=random.choices(unvisited,weights=probabilities)[0]

            route.append(next_node)
            unvisited.remove(next_node)
            current=next_node

        route.append("Depot")

        total_distance=0

        for i in range(len(route)-1):
            total_distance+=distance[route[i],route[i+1]]

        all_routes.append((route,total_distance))

        if total_distance<best_distance:
            best_distance=total_distance
            best_route=route.copy()

    for edge in pheromone:
        pheromone[edge]*=(1-rho)

    for route,cost in all_routes:
        deposit=Q/cost

        for i in range(len(route)-1):
            edge=(route[i],route[i+1])
            pheromone[edge]+=deposit

robot_speed=2
travel_time=best_distance/robot_speed

print("Best Picking Route:")
print(" -> ".join(best_route))

print("Total Distance:",round(best_distance,2),"meters")
print("Estimated Travel Time:",round(travel_time,2),"seconds")
print("ACO Optimization Completed!")
```
