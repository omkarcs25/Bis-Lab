```python

import random
import math

N = 10
MAX_ITER = 50
AREA = 100
RADIUS = 20

# Generate fixed target points
points = [
    (random.randint(0, AREA), random.randint(0, AREA))
    for _ in range(100)
]

# Calculate coverage
def coverage(position):
    x, y = position
    count = 0

    for px, py in points:
        distance = math.sqrt((x-px)**2 + (y-py)**2)

        if distance <= RADIUS:
            count += 1

    return count

# Fitness function
def fitness(position):
    x, y = position
    energy = math.sqrt(x**2 + y**2)

    return coverage(position) - 0.01 * energy

# Initialize cells
cells = [
    [random.uniform(0, AREA), random.uniform(0, AREA)]
    for _ in range(N)
]

# Parallel Cellular Algorithm
for iteration in range(MAX_ITER):

    fitness_values = [fitness(cell) for cell in cells]
    new_cells = []

    for i in range(N):

        left = (i - 1) % N
        right = (i + 1) % N

        neighbours = [left, i, right]

        best = max(neighbours, key=lambda j: fitness_values[j])

        x, y = cells[best]

        new_x = x + random.uniform(-5, 5)
        new_y = y + random.uniform(-5, 5)

        new_x = max(0, min(AREA, new_x))
        new_y = max(0, min(AREA, new_y))

        candidate = [new_x, new_y]

        if fitness(candidate) > fitness_values[i]:
            new_cells.append(candidate)
        else:
            new_cells.append(cells[i])

    cells = new_cells

# Display all sensor positions and coverage
print("\nSENSOR POSITIONS AND COVERAGE\n")

for i, sensor in enumerate(cells):
    x, y = sensor
    covered = coverage(sensor)

    print(f"Sensor {i+1}:")
    print(f"Position: ({x:.2f}, {y:.2f})")
    print(f"Covered Points: {covered}/100")
    print(f"Coverage: {covered}%")
    print(f"Coverage Area: {math.pi * RADIUS**2:.2f} square units")
    print()

# Find best sensor position
best_cell = max(cells, key=fitness)

print("BEST SENSOR POSITION")
print("Position:", best_cell)
print("Covered Points:", coverage(best_cell))
print("Maximum Fitness:", fitness(best_cell))
```
